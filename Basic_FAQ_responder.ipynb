```python
def get_faq_response(query):
    query = query.lower()

    faqs = {
        "timings": "The institute operates from 9 AM to 5 PM, Monday to Friday.",
        "fees": "1200000",
        "contact": "You can reach us at info@institute.edu or call us at (123) 456-7890.",
        "courses": "We offer a wide range of courses in various disciplines, including Computer Science, Engineering, Business, and Arts. Please check our prospectus for more details.",
        "admission process": "The admission process involves an online application, submission of documents, and an entrance exam/interview. Details are available on the admissions page.",
        "library hours": "Our library is open from 8 AM to 8 PM on weekdays and 10 AM to 4 PM on Saturdays.",
        "scholarships": "Yes, we offer various scholarships based on merit and financial need. Please refer to the scholarships section on our website.",
        "location": "The institute is located at 123 University Ave, City, State, ZIP. You can find us on Google Maps.",
        "faculty": "Our faculty comprises highly qualified and experienced professionals in their respective fields.",
        "hostel": "Hostel facilities are available for out-of-town students. Please inquire at the student services office for availability and charges.",
        "exit": "Please use our services again.",
        "branches": "Our branches are located in Nagpur, Mumbai, Pune, Nashik and Sambhajinagar"

    }

    if "timing" in query or "hours" in query:
        return faqs["timings"]
    elif "fee" in query or "cost" in query or "price" in query:
        return faqs["fees"]
    elif "contact" in query or "reach us" in query or "phone" in query or "email" in query:
        return faqs["contact"]
    elif "course" in query or "program" in query or "study" in query:
        return faqs["courses"]
    elif "admission" in query or "apply" in query:
        return faqs["admission process"]
    elif "library" in query:
        return faqs["library hours"]
    elif "scholarship" in query or "aid" in query:
        return faqs["scholarships"]
    elif "location" in query or "address" in query or "where are you" in query:
        return faqs["location"]
    elif "faculty" in query or "teacher" in query or "professor" in query:
        return faqs["faculty"]
    elif "hostel" in query or "accommodation" in query or "dorm" in query:
        return faqs["hostel"]
    elif "exit" in query:
        return faqs["exit"]
    elif "branches" in query:
        return faqs["branches"]
    else:
        return "I'm sorry, I don't have an answer to that question. Please try rephrasing or ask about timings, fees, or contact information."

print("Welcome to the Institute FAQ Chatbot! Type 'exit' to quit.\n")

while True:
    user_query = input("Ask me anything: ")
    if user_query.lower() == 'exit':
        print("Please use our services again.")
        break # Added break to exit the loop

    response = get_faq_response(user_query)
    print(f"Chatbot: {response}")
    print("Hope we helped")
```
